```python
"""
Elevate Labs AI & ML Internship
Task 2: Exploratory Data Analysis (EDA)
Dataset: Titanic-Dataset.csv
"""

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set plotting aesthetics
sns.set_theme(style="whitegrid", palette="pastel")

def load_and_summarize(filepath):
    """Loads the dataset and prints summary statistics, missing values, skewness, and kurtosis."""
    df = pd.read_csv(filepath)
    print("--- Dataset Shape ---")
    print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}\n")

    print("--- Missing Values ---")
    print(df.isnull().sum()[df.isnull().sum() > 0])
    print("\n")

    print("--- Summary Statistics (Numeric) ---")
    numeric_df = df.select_dtypes(include=['int64', 'float64'])
    print(numeric_df.describe().round(2))
    print("\n")

    print("--- Skewness & Kurtosis ---")
    print(f"Skewness:\n{numeric_df.skew().round(2)}\n")
    print(f"Kurtosis:\n{numeric_df.kurtosis().round(2)}\n")

    return df

def generate_basic_visuals(df):
    """Generates and saves histograms and boxplots for Age and Fare."""
    fig, axes = plt.subplots(2, 2, figsize=(16, 12))

    # Histograms
    sns.histplot(df['Age'].dropna(), kde=True, bins=30, ax=axes[0, 0], color='cornflowerblue')
    axes[0, 0].set_title('Age Distribution of Passengers')

    sns.histplot(df['Fare'].dropna(), kde=True, bins=40, ax=axes[0, 1], color='lightcoral')
    axes[0, 1].set_title('Fare Distribution (Right-Skewed)')

    # Boxplots by Survival
    sns.boxplot(x='Survived', y='Age', data=df, ax=axes[1, 0], palette='Set2')
    axes[1, 0].set_title('Age Distribution by Survival Status')
    axes[1, 0].set_xticklabels(['Did Not Survive (0)', 'Survived (1)'])

    sns.boxplot(x='Survived', y='Fare', data=df, ax=axes[1, 1], palette='Set2')
    axes[1, 1].set_title('Fare Paid by Survival Status')
    axes[1, 1].set_xticklabels(['Did Not Survive (0)', 'Survived (1)'])
    axes[1, 1].set_ylim(0, 200) # Cap y-axis to view interquartile range clearly

    plt.tight_layout()
    plt.savefig('histograms_and_boxplots.png', dpi=300)
    plt.close()
    print("[*] Saved basic distributions to 'histograms_and_boxplots.png'")

def generate_relationship_visuals(df):
    """Generates and saves the correlation matrix and pairplot."""
    # Correlation Matrix
    numeric_df = df.select_dtypes(include=['int64', 'float64'])
    corr_matrix = numeric_df.corr()

    plt.figure(figsize=(10, 8))
    sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
    plt.title('Feature Correlation Matrix', fontsize=16)
    plt.tight_layout()
    plt.savefig('correlation_matrix.png', dpi=300)
    plt.close()
    print("[*] Saved correlation matrix to 'correlation_matrix.png'")

    # Pairplot
    pairplot_fig = sns.pairplot(df[['Age', 'Fare', 'Pclass', 'Survived']].dropna(), hue='Survived', palette='husl')
    pairplot_fig.savefig('pairplot.png', dpi=300)
    plt.close()
    print("[*] Saved pairplot to 'pairplot.png'")

def generate_advanced_visuals(df):
    """Generates a multi-plot dashboard for deeper feature-level inferences."""
    fig, axes = plt.subplots(2, 2, figsize=(16, 12))

    # Violin Plot: Age across Pclass and Survival
    sns.violinplot(x="Pclass", y="Age", hue="Survived", data=df, split=True, ax=axes[0, 0], palette="pastel")
    axes[0, 0].set_title('Age Distribution by Ticket Class & Survival')

    # Countplot: Embarked by Survival
    sns.countplot(x="Embarked", hue="Survived", data=df, ax=axes[0, 1], palette="pastel")
    axes[0, 1].set_title('Survival Count by Embarkation Port')

    # Scatter Plot: Age vs Fare
    sns.scatterplot(x="Age", y="Fare", hue="Survived", data=df, alpha=0.7, ax=axes[1, 0], palette="pastel")
    axes[1, 0].set_title('Age vs Fare (Colored by Survival)')
    axes[1, 0].set_ylim(0, 300)

    # KDE Plot: Age Density by Survival
    sns.kdeplot(data=df, x="Age", hue="Survived", fill=True, ax=axes[1, 1], palette="pastel")
    axes[1, 1].set_title('Age Density by Survival')

    plt.tight_layout()
    plt.savefig('advanced_eda_dashboard.png', dpi=300)
    plt.close()
    print("[*] Saved advanced EDA dashboard to 'advanced_eda_dashboard.png'")

if __name__ == "__main__":
    filepath = 'Titanic-Dataset.csv'

    try:
        df = load_and_summarize(filepath)
        generate_basic_visuals(df)
        generate_relationship_visuals(df)
        generate_advanced_visuals(df)
        print("\n[*] EDA Pipeline Completed Successfully.")
    except FileNotFoundError:
        print(f"[!] Error: The file '{filepath}' was not found. Please ensure it is in the same directory.")
```
